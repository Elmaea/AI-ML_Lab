```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate a synthetic 2D dataset with 4 centers
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

# Visualize the raw data
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], s=50, cmap='viridis')
plt.title("Visualizing Raw Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

```python
optimal_k = 3
k_means_model = KMeans(n_clusters=optimal_k, init='k-means++', max_iter=300, n_init=10, random_state=42)
y_kmeans = k_means_model.fit_predict(X)
```

```python
# Plot the clusters and centroids
plt.figure(figsize=(8, 6))

# Plot points belonging to each of the 3 clusters
for i in range(optimal_k):
    plt.scatter(X[y_kmeans == i, 0], X[y_kmeans == i, 1], s=50, label=f'Cluster {i+1}')

# Plot the centroids
plt.scatter(k_means_model.cluster_centers_[:, 0], k_means_model.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroids')

plt.title('k-Means Clustering Results with k=3')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
